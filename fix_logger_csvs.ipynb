```python
import pandas as pd
import os
import glob
```

```python
file_path = "\\\\APVEQUISWEB1\\equis_live\\Consumers\\Transducers"
# only list files modified after a certain date
date_threshold = pd.Timestamp('2026-07-01').timestamp()  # Example date threshold
modified_files = [f for f in os.listdir(file_path) if os.path.getmtime(os.path.join(file_path, f)) > date_threshold]
modified_files
```

```python
new_csvs = [modified_file for modified_file in modified_files if modified_file.endswith('.csv')]
len(new_csvs)
```

```python
csv = new_csvs[35]  # Display the first new CSV file name
df = pd.read_csv(os.path.join(file_path, csv))

mask = df.iloc[:, 0].astype(str).str.contains("Log Name", na=False)
if mask.any():
    row_index = mask[mask].index[0]
    print(row_index)  # Display the row index where "Log Name" is found
    df.iloc[row_index:row_index + 5]
else:
    print("Log Name not found in first column")
df.iloc[row_index:row_index + 5]
```

```python
for i in range(len(new_csvs)):
    csv = new_csvs[i]
    df = pd.read_csv(os.path.join(file_path, csv))
    mask = df.iloc[:, 0].astype(str).str.contains("Log Name", na=False)
    if mask.any():
        row_index = mask[mask].index[0]
        print(row_index)  # Display the row index where "Log Name" is found
        df.iloc[row_index:row_index + 5]
    else:
        print("Log Name not found in first column")
    df.iloc[row_index:row_index + 5]
    df2 = df.copy()
    df2.iloc[row_index, 0] = df.iloc[row_index, 0].replace("-Log", "").replace("_", "").replace("2026-05-04", "").replace("2026-05", "").replace("- -", " - ").replace(" -  ", " - ").replace("-05","")
    df2.iloc[row_index:row_index + 5]
    df2.to_csv(os.path.join("output", f"{csv}"), index=False)
```
