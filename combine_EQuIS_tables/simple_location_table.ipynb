```python
import pandas as pd
```

```python
def get_location_table(root):
    dt_coordinate = pd.read_excel(root + r"\dt_coordinate.xlsx")
    table = dt_coordinate[["sys_loc_code", "y_coord", "x_coord", 'elev', 'coord_type_code']]
    table_xy = table[table['coord_type_code'].str.contains("SP")]
    table_xy = table_xy.rename(columns={"y_coord": "Northing", "x_coord": "Easting", "elev": "elevation"})   
    table_ll = table[table['coord_type_code'].str.contains("LAT")]
    table_ll = table_ll.rename(columns={"y_coord": "Latitude", "x_coord": "Longitude", "elev": "elevation"})
    table_xy = table_xy[["sys_loc_code", "Northing", "Easting", "elevation"]]
    table_ll = table_ll[["sys_loc_code", "Latitude", "Longitude", "elevation"]]
    final = pd.merge(table_xy, table_ll, on=["sys_loc_code", "elevation"], how="outer")
    final = final.rename(columns={"sys_loc_code": "Location"})
    return final
```

```python
root = r"C:\Users\ageglio\OneDrive - TRC\Documents\My EQuIS Work\Mdn01\CCR-Monroe Fly Ash Basin"
# root = r"C:\Users\ageglio\OneDrive - TRC\Documents\My EQuIS Work\Consumers\JR Whiting - Erie"

final = get_location_table(root)
final.to_excel("CCR-Menroe_FAB-BAI_Locations.xlsx", index=False)
# final.to_excel("JRWhiting_Locations.xlsx", index=False)
```
