```python
def merge_water_level_measure_datum(
    water_level,
    measure_datum,
    date_column_name="measure_date"
):
    water_level = water_level.copy()
    measure_datum = measure_datum.copy()

    # --- Normalize AND force identical datetime resolution ---
    water_level[date_column_name] = (
        pd.to_datetime(water_level[date_column_name], errors="coerce")
        .dt.normalize()
        .astype("datetime64[ns]")
    )

    measure_datum["start_date"] = (
        pd.to_datetime(measure_datum["start_date"], errors="coerce")
        .dt.normalize()
        .astype("datetime64[ns]")
    )

    water_level = water_level.dropna(subset=[date_column_name])
    measure_datum = measure_datum.dropna(subset=["start_date"])

    # IMPORTANT: sort by time first, then by group key
    water_level = (
        water_level
        .sort_values([date_column_name, "sys_loc_code"], kind="mergesort")
        .reset_index(drop=True)
    )

    measure_datum = (
        measure_datum
        .sort_values(["start_date", "sys_loc_code"], kind="mergesort")
        .reset_index(drop=True)
    )

    merged = pd.merge_asof(
        water_level,
        measure_datum,
        left_on=date_column_name,
        right_on="start_date",
        by="sys_loc_code",
        direction="backward"
    )

    return merged
```

```python
# ---  Merge the earliest water level with the measure datum to get the elevation of the measurement ---
merged_earliest = merge_water_level_measure_datum(dt_water_level_earliest, dt_measure_datum2, date_column_name="earliest_measurement_date")
merged_earliest = merged_earliest.rename(
    columns={"start_date":"earliest_measure_datum_start_date", "datum_value":"earliest_measure_datum_value"}
)
merged_Q12026 = merge_water_level_measure_datum(dt_water_level_Q12026, dt_measure_datum2, date_column_name="Q12026_measurement_date")
merged_Q12026 = merged_Q12026.rename(
    columns={"start_date":"Q12026_measure_datum_start_date", "datum_value":"Q12026_measure_datum_value"}
)
```

```python
# --- 1. Merge all dataframes on sys_loc_code ---

df_merged = (
    dt_location2
    .merge(dt_coordinate2, on="sys_loc_code", how="left")
    .merge(dt_well_segment2, on="sys_loc_code", how="left")
    .merge(merged_earliest, on="sys_loc_code", how="left")
    .merge(merged_Q12026, on="sys_loc_code", how="left")
    .merge(dt_well2, on="sys_loc_code", how="left")
    .merge(df_cat, on="sys_loc_code", how="left")
)


# --- 2. Create a sort order based on your original lists ---

ordered_list = (
    a_karn_bottom_ash_pond +
    b_karn_bottom_ash_pond_supplemental +
    c_dek_bottom_ash_and_karn_lined_impoundment +
    d_karn_lined_impoundment +
    e_karn_landfill_perimeter_dike +
    f_karn_gsi_wells +
    g_karn_solid_waste_disposal_area +
    h_karn_leachate_headwells +
    i_karn_landfill_swl_only +
    j_karn_temporary_wells +
    k_weadock_background +
    l_weadock_landfill_downgradient_outside_swl +
    m_weadock_landfill_swl_only_inside_swl +
    n_weadock_landfill_leachate_headwells +
    o_weadock_bottom_ash_pond_downgradient_closed +
    p_other_weadock_site_wells +
    q_weadock_bedrock_wells
)

# Ensure it's unique
ordered_list = list(dict.fromkeys(ordered_list))

# Map each well to its sort index
sort_map = {well: i for i, well in enumerate(ordered_list)}

# --- 3. Filter merged dataframe to only wells in all_wells ---

df_filtered = df_merged[df_merged["sys_loc_code"].isin(ordered_list)].copy()

# --- 4. Sort by sort map
df_filtered["sort_order"] = df_filtered["sys_loc_code"].map(sort_map)
```

```python
# --- Final output perform calculations ---
df_final = df_filtered.copy()
df_final = df_final[df_final.loc_type != "Decommissioned Well"]
df_final["scr_top_elev"] = df_final.elev - df_final.start_depth
df_final["scr_bot_elev"] = df_final.elev - df_final.end_depth
df_final["bot_elev"] = df_final.elev - df_final.depth_of_well

# Measure Bottom Elevations
df_final["Earliest_meas_bot_elev"] = df_final.earliest_measure_datum_value - df_final.earliest_measured_depth_of_well
df_final["Q12026_Bottom_Elevation"] = df_final.Q12026_measure_datum_value - df_final.Q12026_measured_depth_of_well


df_final["Depth_to_Bottom_Difference_(1)"] = df_final["bot_elev"] - df_final["Q12026_Bottom_Elevation"]
df_final["Depth_to_Bottom_Difference_(2)"] = df_final["Earliest_meas_bot_elev"] - df_final["Q12026_Bottom_Elevation"]
df_final["to1"] = "to"
df_final["to2"] = "to"
df_final = df_final.rename(columns={"elev":"GS_Elev", "datum_value":"TOC_Elevation", "remark":"screen_remark", 
                                    "start_depth":"scr_top", "end_depth":"scr_bottom"})

# re-order
df_final = df_final.sort_values("sort_order")
```
